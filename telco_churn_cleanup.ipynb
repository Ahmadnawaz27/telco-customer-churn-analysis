```python
import pandas as pd

df = pd.read_excel('Telco_customer_churn.xlsx')
print(df.shape)          # 7043 rows, 33 columns
print(df.dtypes)         # check data types
df.head(3)
```

```python
drop_cols = ['Count', 'Lat Long', 'Zip Code', 'Churn Value']
df = df.drop(columns=drop_cols)
```

```python
# Churn Reason is null for non-churners, that's correct, just fill it
df['Churn Reason'] = df['Churn Reason'].fillna('No Churn')
```

```python
def tenure_band(months):
    if months <= 12: return '0–12 months'
    elif months <= 24: return '13–24 months'
    elif months <= 48: return '25–48 months'
    else: return '49–72 months'

df['Tenure Band'] = df['Tenure Months'].apply(tenure_band)
```

```python
df.to_csv('telco_churn_clean.csv', index=False)
print('Done —', df.shape)
```
